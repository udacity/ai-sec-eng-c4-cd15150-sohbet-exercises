## Reference Notes

A few specification details for cross-referencing the EU AI Act citations used in this lesson.

- **GPAI tier provenance.** Articles 51-55 (GPAI) were introduced in Regulation (EU) 2024/1689 when adopted in June 2024. The substantive GPAI obligations became applicable on 2 August 2025. Regulation (EU) 2026/1744 deferred general applicability of the wider high-risk regime to 2 December 2027 for stand-alone Annex III systems and 2 August 2028 for high-risk AI embedded in products already covered by EU product-safety law; those dates concern the wider regime, not the introduction of GPAI.
- **Article 5 workplace emotion recognition.** Article 5(1)(f) prohibits emotion-recognition systems in the workplace and in educational institutions, except for medical or safety reasons. The EmotionSense HR edge case sits squarely against this prohibition.
- **High-risk indicator threshold.** The "require at least 2 high-risk indicators" heuristic used here is a pedagogical guardrail to avoid over-classification, not a statutory rule. The statutory test is Annex III membership; the multi-indicator check approximates that for the systems in this lab.
